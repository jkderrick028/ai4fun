```python
df = pd.read_csv(data_file)
df 
```

```python
df.shape 
```

```python
df_new = df.dropna(inplace=False)
df_new 
```

```python
df_new.shape 
```

```python
df_new = df.fillna(0, inplace=False) 
df_new 
```

```python
df.fillna({'Calories': 0}, inplace=False) 
```

```python
df['Calories'].mean() 
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='mixed')
df 
```

```python
df.dropna(subset=['Date'], inplace=True) 
```

```python
df 
```

```python
df.loc[7, 'Duration'] = 45 
df 
```

```python
for x in df.index:
    if df.loc[x, 'Maxpulse'] > 120:
        df.loc[x, 'Maxpulse'] = 120 
df 
```

```python
for x in df.index:
    if df.loc[x, 'Maxpulse'] > 130:
        df.drop(x, inplace=True) 
df 
```

```python
df.duplicated() 
```

```python
df.drop_duplicates(inplace=False) 
```

```python
df.corr() 
```

```python
df.plot() 
```

```python
df.plot(kind='scatter', x='Maxpulse', y='Calories') 
```

```python
df.groupby('Duration')['Calories'].mean() 
```
